```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt

# Ensure we can import from the main package
sys.path.append(os.path.abspath(".."))

from polymarket_sim.research.markov_analysis import MarkovAnalyzer
from polymarket_sim.research.monte_carlo import MonteCarloSimulator

%matplotlib inline
plt.style.use('seaborn-v0_8-darkgrid')
```

## 1. Load Data
Load your trade history or price data here.

```python
# Example: Load trades from a CSV (you can export your portfolio to CSV)
# df = pd.read_csv("../data/trades.csv")

# For demo purposes, let's create some dummy data representing a winning strategy
import numpy as np
np.random.seed(42)
trades = np.random.normal(loc=5, scale=20, size=100) # Mean profit $5, SD $20
cumulative = np.cumsum(trades)

plt.figure(figsize=(10,4))
plt.plot(cumulative)
plt.title("Historical Equity Curve")
plt.show()
```

## 2. Markov Chain Analysis (Market Regime)
Determine if the market is trending or mean-reverting.

```python
# Example price series
prices = [0.50, 0.52, 0.53, 0.55, 0.54, 0.52, 0.50, 0.48, 0.49]

analyzer = MarkovAnalyzer(prices, threshold=0.01)
matrix = analyzer.analyze()
regime = analyzer.get_regime()

print(f"Detected Regime: {regime}")
pd.DataFrame(matrix).T # Transpose for readability (Row=Current, Col=Next)
```

## 3. Monte Carlo Simulation
What if the order of trades was different? What is the risk of ruin?

```python
mc = MonteCarloSimulator(trades, initial_capital=1000)
sim_results = mc.run(num_sims=5000, trades_per_sim=200)
stats = mc.get_stats()

print("Risk of Ruin:", f"{stats['risk_of_ruin_pct']:.2f}%")
print("VaR 95%:", f"${stats['VaR_95']:.2f}")

plt.figure(figsize=(12, 6))
plt.hist(sim_results[:, -1], bins=50, alpha=0.7, color='purple')
plt.title("Distribution of Final Equity (5000 Simulations)")
plt.axvline(stats['VaR_95'], color='r', linestyle='--', label='95% VaR')
plt.legend()
plt.show()
```
