### 📝 Các công thức quan trọng (Key Formulas)

*   **Sampling Theorem (Định lý lấy mẫu Nyquist):**
    $$f_s \ge 2f_{max}$$
    *   **When to use**: Determining the minimum sampling rate required to avoid aliasing when converting analog signals to digital.
    *   **Variables**:
        *   $f_s$: Sampling frequency (Hz)
        *   $f_{max}$: Maximum frequency component in the analog signal (Hz)

*   **Energy of Discrete Signal (Năng lượng):**
    $$E_x = \sum_{n=-\infty}^{\infty} |x(n)|^2$$
    *   **When to use**: Calculating the total energy of a finite-duration or decaying signal. If $0 < E_x < \infty$, it's an **Energy Signal**.
    *   **Variables**:
        *   $x(n)$: Discrete signal amplitude at index $n$

*   **Power of Discrete Signal (Công suất):**
    $$P_x = \lim_{N\to\infty} \frac{1}{2N+1} \sum_{n=-N}^{N} |x(n)|^2$$
    *   **When to use**: Calculating the average power of infinite-duration, periodic signals. If $0 < P_x < \infty$, it's a **Power Signal**.
    *   **Variables**:
        *   $2N+1$: Total number of samples in the interval

*   **Convolution Sum (Tích chập):**
    $$y(n) = x(n) * h(n) = \sum_{k=-\infty}^{\infty} x(k)h(n-k)$$
    *   **When to use**: Finding the output $y(n)$ of an LTI system given input $x(n)$ and impulse response $h(n)$.
    *   **Variables**:
        *   $h(n)$: Impulse response of the system
        *   $k$: Dummy variable for summation

*   **BIBO Stability Condition:**
    $$\sum_{n=-\infty}^{\infty} |h(n)| < \infty$$
    *   **When to use**: Checking if an LTI system is Stable (Bounded Input produces Bounded Output).
    *   **Variables**:
        *   $|h(n)|$: Absolute value of the impulse response

### 📝 Các công thức quan trọng (Key Formulas)

*   **Z-Transform Definition:**
    $$X(z) = \mathcal{Z}\{x(n)\} = \sum_{n=-\infty}^{\infty} x(n)z^{-n}$$
    *   **When to use**: Converting a discrete time-domain signal $x(n)$ to the Z-domain $X(z)$ for analysis (stability, frequency response).
    *   **Variables**:
        *   $z$: Complex variable ($z = r e^{j\omega}$)
        *   $ROC$: Region of Convergence (values of $z$ where sum converges)

*   **Time Shifting Property:**
    $$x(n-k) \leftrightarrow z^{-k}X(z)$$
    *   **When to use**: Finding the Z-transform of a delayed or advanced signal. Essential for solving difference equations.
    *   **Variables**:
        *   $k$: Number of delay steps (integer)

*   **Convolution Property:**
    $$x_1(n) * x_2(n) \leftrightarrow X_1(z)X_2(z)$$
    *   **When to use**: Calculating the output response $Y(z) = H(z)X(z)$ in Z-domain instead of time-domain convolution.

*   **Initial Value Theorem (Casual Signal):**
    $$x(0) = \lim_{z\to\infty} X(z)$$
    *   **When to use**: Finding the first sample $x(0)$ directly from $X(z)$ without performing full inverse transform.

*   **Final Value Theorem:**
    $$x(\infty) = \lim_{z\to 1} (z-1)X(z)$$
    *   **When to use**: Finding the steady-state value $x(\infty)$ of the signal (only if all poles of $(z-1)X(z)$ are inside unit circle).


## Chương 3 & 4: Miền tần số

- **DFT**: Biến đổi Fourier rời rạc
- **FFT**: Thuật toán biến đổi nhanh (Cơ số 2)

### 📝 Các công thức quan trọng (Key Formulas)

*   **DTFT Definition (Biến đổi Fourier thời gian rời rạc):**
    $$X(e^{j\omega}) = \sum_{n=-\infty}^{\infty} x(n) e^{-j\omega n}$$
    *   **When to use**: Analyzing the frequency content of an aperiodic discrete signal. Spectrum is continuous and periodic.
    *   **Variables**:
        *   $\omega$: Frequency in radians ($-\pi$ to $\pi$)

*   **DFT Definition (Biến đổi Fourier rời rạc):**
    $$X(k) = \sum_{n=0}^{N-1} x(n) W_N^{kn}, \quad W_N = e^{-j\frac{2\pi}{N}}$$
    *   **When to use**: Computational version of Fourier Transform for finite-length signals (computers use this via FFT).
    *   **Variables**:
        *   $N$: Length of the DFT (number of frequency points)
        *   $k$: Frequency index ($0$ to $N-1$)
        *   $W_N$: Twiddle factor

*   **IDFT:**
    $$x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k) W_N^{-kn}$$
    *   **When to use**: Reconstructing the time-domain signal from its frequency samples.

*   **Circular Convolution:**
    $$x_1(n) \circledast x_2(n) \leftrightarrow X_1(k)X_2(k)$$
    *   **When to use**: Computing the result of input passing through a system using DFT/FFT. Note: Linear convolution requires padding to avoid aliasing.


## Chương 5: Bộ lọc số

- **Bộ lọc FIR**: Thiết kế bằng phương pháp cửa sổ
- **Các loại cửa sổ**:
  - Hình chữ nhật
  - Hamming
  - Hanning
  - Blackman
  - Kaiser

### 📝 Các công thức quan trọng (Key Formulas)

*   **Difference Equation (Phương trình sai phân):**
    $$y(n) = -\sum_{k=1}^{N} a_k y(n-k) + \sum_{k=0}^{M} b_k x(n-k)$$
    *   **When to use**: Implementing the filter in time-domain (software/hardware direct realization).
    *   **Variables**:
        *   $a_k$: Feedback coefficients (poles)
        *   $b_k$: Feedforward coefficients (zeros)

*   **System Function (Hàm truyền đạt):**
    $$H(z) = \frac{Y(z)}{X(z)} = \frac{\sum_{k=0}^{M} b_k z^{-k}}{1 + \sum_{k=1}^{N} a_k z^{-k}}$$
    *   **When to use**: Designing filters, analyzing stability (poles), and frequency response (zeros/poles on Z-plane).

*   **FIR Filter:** (No feedback, $a_k = 0$)
    $$H(z) = \sum_{k=0}^{M} b_k z^{-k}$$
    *   **When to use**: When linear phase is required (e.g., audio processing) and stability must be guaranteed.


# Tóm tắt nội dung bài giảng (Slide_XLTHS_1.pdf)

## 1. Tín hiệu và Hệ thống rời rạc
- **Tín hiệu**: Đại lượng vật lý mang tin tức, biến thiên theo thời gian/không gian. 
- **Phân loại**: Tương tự (Analog) vs Số (Digital). Ưu điểm của xử lý số: linh hoạt, lưu trữ dễ dàng, chi phí thấp, độ chính xác cao.
- **ADC/DAC**: 
    - **Lấy mẫu (Sampling)**: Rời rạc hóa theo thời gian. Định lý Nyquist: $f_s \ge 2f_{max}$.
    - **Lượng tử hóa (Quantization)**: Rời rạc hóa biên độ.
    - **Mã hóa (Coding)**: Chuyển giá trị lượng tử sang nhị phân.
- **Tín hiệu cơ bản**: Xung đơn vị $\delta(n)$, bước nhảy đơn vị $u(n)$, dốc đơn vị $r(n)$, chữ nhật $rect_N(n)$, sin/cos.
- **Các phép toán**: Cộng, nhân, dịch (trễ/sớm), tích chập $y(n) = x(n) * h(n)$.
- **Hệ thống**: Tuyến tính (Linear), Bất biến (Time-Invariant), Nhân quả (Causal), Ổn định (Stable).
    - **LTI (Linear Time-Invariant)**: Đặc trưng bởi đáp ứng xung $h(n)$. $y(n) = \sum x(k)h(n-k)$.

## 2. Biến đổi Z (Z-Transform)
- **Định nghĩa**: $X(z) = \sum_{n=-\infty}^{\infty} x(n)z^{-n}$.
- **ROC (Miền hội tụ)**: Vùng giá trị $z$ nơi chuỗi hội tụ. Quyết định tính ổn định và nhân quả.
- **Điểm cực (Poles) & Điểm không (Zeros)**: Vị trí của chúng trên mặt phẳng phức Z quyết định đặc tính tần số của hệ thống.
- **Tính chất**: Tuyến tính, dịch thời gian, tỷ lệ, đạo hàm, tích chập ($x_1(n)*x_2(n) \leftrightarrow X_1(z)X_2(z)$).
- **Biến đổi ngược (IZT)**: Thặng dư, chia đa thức, phân thức tối giản.
- **Hàm truyền đạt**: $H(z) = Y(z)/X(z) = \sum b_k z^{-k} / (1 - \sum a_k z^{-k})$.

## 3 & 4. Miền tần số (Frequency Domain)
- **DTFT (Discrete-Time Fourier Transform)**: Phổ của tín hiệu rời rạc không tuần hoàn. $X(e^{j\omega})$ liên tục, tuần hoàn $2\pi$.
- **DFT (Discrete Fourier Transform)**: Biến đổi Fourier rời rạc cho tín hiệu hữu hạn chiều dài N. $X(k) = \sum_{n=0}^{N-1} x(n)W_N^{kn}$.
- **FFT (Fast Fourier Transform)**: Thuật toán tính nhanh DFT (Cơ số 2, phân chia thời gian/tần số). Độ phức tạp giảm từ $O(N^2)$ xuống $O(N\log N)$.
- **Quan hệ**: DFT là mẫu rời rạc của DTFT. Lấy mẫu miền tần số.

## 5. Bộ lọc số (Digital Filters)
- **FIR (Finite Impulse Response)**:
    - Đáp ứng xung hữu hạn, luôn ổn định, pha tuyến tính.
    - Thiết kế: Phương pháp cửa sổ (Rectangular, Bartlett, Hanning, Hamming, Blackman, Kaiser), Lấy mẫu tần số, Tối ưu (Remez).
- **IIR (Infinite Impulse Response)**:
    - Đáp ứng xung vô hạn, có thể không ổn định, pha phi tuyến, hiệu quả hơn FIR về bậc (ít tính toán hơn cho cùng độ dốc).
    - Thiết kế: Dựa trên bộ lọc tương tự (Butterworth, Chebyshev I/II, Elliptic).
    - Biến đổi: Tương đương vi phân, Xung bất biến, Biến đổi song tuyến (Bilinear Transform).


# Chapter 1 Simulation: Sampling & Aliasing

This simulation demonstrates **Aliasing**. When a signal is sampled at a frequency $f_s$ that is lower than twice its maximum frequency ($f_s < 2f_{max}$), high-frequency components can appear as lower-frequency "aliases".

**Example**:
- **Original Signal**: $9Hz$ Cosine Wave (Cyan)
- **Sampling Rate**: $10Hz$ (Yellow dots)
- **Aliased Signal**: The samples perfectly fit a $1Hz$ Cosine Wave (Magenta dashed), making the 9Hz signal indistinguishable from 1Hz.

```python
import numpy as np
import matplotlib.pyplot as plt

# Settings for '3b1b' style dark plot
plt.style.use('dark_background')
plt.rcParams.update({"grid.color": "gray", "grid.alpha": 0.3})

# Parameters
f0 = 9      # Signal frequency (Hz)
fs = 10     # Sampling frequency (Hz)
duration = 1

# Continuous time (high resolution)
t_cont = np.linspace(0, duration, 1000)
x_cont = np.cos(2 * np.pi * f0 * t_cont)

# Discrete time (sampling instants)
n = np.arange(0, duration * fs + 1)
t_sample = n / fs
x_sample = np.cos(2 * np.pi * f0 * t_sample)

# Alias frequency calculation
f_alias = abs(f0 - fs)  # |9 - 10| = 1 Hz
x_alias = np.cos(2 * np.pi * f_alias * t_cont)

# Plotting
fig, ax = plt.subplots(figsize=(10, 6))

# 1. Original Continuous Signal
ax.plot(t_cont, x_cont, color='cyan', label=f'Original: {f0} Hz', linewidth=2, alpha=0.8)

# 2. Samples
ax.stem(t_sample, x_sample, linefmt='y:', markerfmt='yo', basefmt=' ', label=f'Samples: {fs} Hz')

# 3. Aliased Signal
ax.plot(t_cont, x_alias, color='magenta', linestyle='--', label=f'Alias: {f_alias} Hz', linewidth=2)

# Styling
ax.set_title(f'Aliasing Effect: {f0}Hz sampled at {fs}Hz looks like {f_alias}Hz', fontsize=16, color='white')
ax.set_xlabel('Time (s)', fontsize=12)
ax.set_ylabel('Amplitude', fontsize=12)
ax.legend(loc='upper right', frameon=True, facecolor='black', edgecolor='gray')
ax.grid(True)

plt.tight_layout()
plt.show()
```

# Chapter 2 Simulation: Z-Plane & Frequency Response

This simulation visualizes the relationship between the **Z-Plane (Poles/Zeros)** and the **Frequency Response** of a system.

**Concept**:
- The Z-plane shows the complex plane where we place poles (x) and zeros (o).
- The **Unit Circle** ($|z|=1$) corresponds to the frequency response (from $\omega=0$ to $2\pi$).
- A **Pole** near the unit circle creates a **Peak** (Resonance) at that frequency.
- A **Zero** near the unit circle creates a **Valley** (Notch) at that frequency.

**Example**:
- Pole at $p = 0.9 e^{j\pi/4}$ (Close to unit circle at angle $45^\circ$)
- Zero at $z = 0$ (Origin)

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
theta = np.pi / 4  # 45 degrees
r = 0.9            # Radius (close to 1)
pole = r * np.exp(1j * theta)
zero = 0

# Create figure
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# --- 1. Z-Plane Plot ---
# Draw Unit Circle
unit_circle = plt.Circle((0, 0), 1, color='white', fill=False, linestyle='--', alpha=0.5)
ax1.add_patch(unit_circle)

# Plot Axes
ax1.axhline(0, color='gray', lw=1)
ax1.axvline(0, color='gray', lw=1)

# Plot Pole and Zero
ax1.plot(np.real(pole), np.imag(pole), 'rx', markersize=12, markeredgewidth=3, label='Pole')
ax1.plot(np.real(zero), np.imag(zero), 'go', markersize=12, markerfacecolor='none', markeredgewidth=3, label='Zero')

# Styling
ax1.set_xlim(-1.5, 1.5)
ax1.set_ylim(-1.5, 1.5)
ax1.set_aspect('equal')
ax1.set_title("Z-Plane (Pole-Zero Plot)", fontsize=14, color='white')
ax1.set_xlabel("Real", fontsize=12)
ax1.set_ylabel("Imaginary", fontsize=12)
ax1.legend(facecolor='black', edgecolor='gray')

# --- 2. Frequency Response Plot ---
w = np.linspace(0, np.pi, 500)  # Frequency from 0 to pi
z = np.exp(1j * w)              # Points on unit circle

# Calculate H(z) = (z - zero) / (z - pole)
H = (z - zero) / (z - pole)
mag_H = np.abs(H)

# Plot Magnitude
ax2.plot(w, mag_H, color='yellow', linewidth=2)

# Highlight the frequency corresponding to the pole
ax2.axvline(theta, color='red', linestyle='--', alpha=0.7, label=f'Pole Angle = {theta:.2f} rad')

# Styling
ax2.set_title("Frequency Response Magnitude |H(e^jw)|", fontsize=14, color='white')
ax2.set_xlabel("Frequency (rad/sample)", fontsize=12)
ax2.set_ylabel("Magnitude", fontsize=12)
ax2.set_xticks([0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi])
ax2.set_xticklabels(['0', r'$\pi/4$', r'$\pi/2$', r'$3\pi/4$', r'$\pi$'])
ax2.grid(True)
ax2.legend(facecolor='black', edgecolor='gray')

plt.tight_layout()
plt.show()
```

# Chapter 3 Simulation: DTFT & Periodicity

This simulation illustrates the **Discrete-Time Fourier Transform (DTFT)**.

**Concept**:
- The spectrum of a discrete-time signal is always **periodic** with period $2\pi$.
- We calculate the DTFT of a simple rectangular pulse $x(n) = 1$ for $0 \le n < M$.
- The result is a sinc-like function (Dirichlet function) in frequency, repeating every $2\pi$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Signal parameters
M = 5
n = np.arange(0, 20)           # Time indices
x = np.where(n < M, 1, 0)      # Rectangular pulse of length M

# DTFT Calculation (Continuous frequency w)
w = np.linspace(-3 * np.pi, 3 * np.pi, 1000)
# DTFT formula: X(w) = sum(x[n] * exp(-j*w*n))
X = np.array([np.sum(x * np.exp(-1j * omega * n)) for omega in w])
mag_X = np.abs(X)

# Plotting
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))

# 1. Time Domain Signal
ax1.stem(n, x, linefmt='c-', markerfmt='co', basefmt=' ')
ax1.set_title("Discrete Time Signal x(n) (Rectangular Pulse)", fontsize=14, color='white')
ax1.set_xlabel("n", fontsize=12)
ax1.set_ylabel("x(n)", fontsize=12)
ax1.grid(True)

# 2. Frequency Domain (DTFT)
ax2.plot(w, mag_X, color='lime', linewidth=2)
ax2.set_title("DTFT Magnitude |X(e^jw)| (Periodic!)", fontsize=14, color='white')
ax2.set_xlabel("Frequency (rad/sample)", fontsize=12)
ax2.set_ylabel("Magnitude", fontsize=12)
ax2.grid(True)

# Highlight periods
ticks = [-2*np.pi, 0, 2*np.pi]
tick_labels = [r'$-2\pi$', '0', r'$2\pi$']
ax2.set_xticks(ticks)
ax2.set_xticklabels(tick_labels)

for val in ticks:
    ax2.axvline(val, color='white', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

# Chapter 4 Simulation: DFT vs DTFT (Sampling in Frequency)

This simulation shows how the **DFT** (Discrete Fourier Transform) samples the continuous **DTFT** spectrum.

**Concept**:
- The DTFT is continuous in frequency.
- The DFT is discrete, calculated at $N$ evenly spaced points around the unit circle.
- If we take an N-point DFT of a signal length L (where $N \ge L$), the DFT values lie exactly on the DTFT curve.

```python
import numpy as np
import matplotlib.pyplot as plt

# Signal: Rectangular pulse again
L = 5
N = 16  # DFT point count (N > L)
n = np.arange(0, N)
x = np.where(n < L, 1, 0)

# 1. Calculate DTFT (Continuous)
w_cont = np.linspace(0, 2 * np.pi, 500)
X_dtft = np.array([np.sum(x * np.exp(-1j * w * n)) for w in w_cont])
mag_dtft = np.abs(X_dtft)

# 2. Calculate DFT (Discrete)
k = np.arange(N)
w_k = 2 * np.pi * k / N
X_dft = np.fft.fft(x, N)
mag_dft = np.abs(X_dft)

# Plotting
fig, ax = plt.subplots(figsize=(12, 6))

# Plot DTFT
ax.plot(w_cont, mag_dtft, color='cyan', label='DTFT (Continuous)', linewidth=2)

# Plot DFT samples
ax.stem(w_k, mag_dft, linefmt='m--', markerfmt='mo', basefmt=' ', label=f'{N}-Point DFT (Samples)')

# Styling
ax.set_title(f'DFT samples the DTFT: Signal Length L={L}, DFT Size N={N}', fontsize=16, color='white')
ax.set_xlabel('Frequency (rad/sample)', fontsize=12)
ax.set_ylabel('Magnitude', fontsize=12)
ax.set_xticks([0, np.pi, 2*np.pi])
ax.set_xticklabels(['0', r'$\pi$', r'$2\pi$'])
ax.grid(True)
ax.legend(facecolor='black', edgecolor='gray')

plt.tight_layout()
plt.show()
```

# Chapter 5 Simulation: FIR vs IIR Impulse Response

This simulation compares the **Impulse Response $h(n)$** of Finite Impulse Response (FIR) and Infinite Impulse Response (IIR) filters.

**Comparison**:
- **FIR Filter**: Impulse response is finite (becomes exactly zero after some time). Example: Moving Average.
- **IIR Filter**: Impulse response is infinite (decays but theoretically never reaches absolute zero). Example: Recursive Decay.

```python
import numpy as np
import matplotlib.pyplot as plt

# Time indices
n = np.arange(0, 20)

# 1. FIR Filter Impulse Response (Moving Average of 5 points)
# h(n) = 0.2 for n=0..4, else 0
h_fir = np.where(n < 5, 0.2, 0)

# 2. IIR Filter Impulse Response (Recursive: y(n) = 0.8*y(n-1) + x(n))
# h(n) = (0.8)^n * u(n)
h_iir = (0.8) ** n

# Plotting
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Plot FIR
ax1.stem(n, h_fir, linefmt='g-', markerfmt='go', basefmt=' ')
ax1.set_title("FIR Impulse Response (Finite)", fontsize=14, color='white')
ax1.set_xlabel("n")
ax1.set_ylim(0, 1)
ax1.grid(True)

# Plot IIR
ax2.stem(n, h_iir, linefmt='r-', markerfmt='ro', basefmt=' ')
ax2.set_title("IIR Impulse Response (Infinite Decay)", fontsize=14, color='white')
ax2.set_xlabel("n")
ax2.set_ylim(0, 1)
ax2.grid(True)

plt.tight_layout()
plt.show()
```
