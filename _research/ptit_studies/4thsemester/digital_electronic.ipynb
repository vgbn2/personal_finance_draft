# TỔNG HỢP KIẾN THỨC KỸ THUẬT SỐ (DIGITAL ELECTRONICS)

Tài liệu tóm tắt các công thức (formulas) và phương pháp giải bài tập (methods) cốt lõi dựa trên học phần Kỹ Thuật Số.

---

## 1. Hệ Đếm & Mã Hóa (Number Systems & Coding)

### Công thức (Formulas)
- **Chuyển đổi cơ số (Base $r$ sang Thập phân 10):**
  $N_{10} = d_{n-1}r^{n-1} + ... + d_1r^1 + d_0r^0 + d_{-1}r^{-1} + ... + d_{-m}r^{-m}$
- **Số bù 1 (1's complement):** Đảo toàn bộ các bit (0 $\rightarrow$ 1, 1 $\rightarrow$ 0).
- **Số bù 2 (2's complement):** Bù 1 + 1. Dùng để biểu diễn số âm và thực hiện phép trừ thông qua phép cộng.
- **Mã BCD (Binary Coded Decimal):** Dùng 4 bit nhị phân biểu diễn 1 chữ số thập phân (từ $0000$ đến $1001$).
- **Mã Gray:** Các từ mã liền kề chỉ khác nhau duy nhất 1 bit. Hữu ích cho Bìa Karnaugh.
- **Chuyển đổi Nhị phân sang Gray:** 
  $G_n = B_n$ (bit MSB giữ nguyên)  
  $G_i = B_{i+1} \oplus B_i$ (các bit còn lại)
- **Chuyển đổi Gray sang Nhị phân:**
  $B_n = G_n$ (bit MSB giữ nguyên)  
  $B_i = B_{i+1} \oplus G_i$

### Phương pháp giải bài tập (Methods)
- **Thập phân $\rightarrow$ Nhị phân / Bát phân / Thập lục phân:** 
  - *Phần nguyên:* Chia liên tiếp cho cơ số (2, 8, 16) và lấy số dư ngược từ dưới lên.
  - *Phần phân số:* Nhân liên tiếp với cơ số, lấy phần nguyên từ trên xuống.
- **Thực hiện phép trừ dùng số bù 2:** Để thực hiện $A - B$, tính $A + (\text{Bù 2 của } B)$. Nếu tổng có bit nhớ (carry) tràn ra, bỏ qua bit đó (kết quả dương). Nếu không có bit nhớ tràn ra, kết quả là âm, bù 2 lại kết quả để lấy độ lớn.

## 2. Đại Số Boole & Cổng Logic (Boolean Algebra & Logic Gates)

### Công thức (Formulas)
#### Các định lý cơ bản:
1. **Định luật De Morgan:** 
   - $\overline{A + B} = \overline{A} \cdot \overline{B}$
   - $\overline{A \cdot B} = \overline{A} + \overline{B}$
2. **Luật nuốt (Absorption):**
   - $A + AB = A$
   - $A(A + B) = A$
   - $A + \overline{A}B = A + B$
3. **Luật phân phối (Distributive):**
   - $A + BC = (A+B)(A+C)$
4. **Cổng XOR và XNOR:**
   - XOR: $A \oplus B = \overline{A}B + A\overline{B}$
   - XNOR: $A \odot B = AB + \overline{A}\ \overline{B}$

### Chuẩn hóa (Canonical Forms)
- **SOP (Tổng các Tích - Sum of Products):** Hàm là tổng các Minterms ($m_i$). Minterm = tích các biến (biến mang dấu phẩy nếu giá trị = 0, không phẩy nếu = 1).
- **POS (Tích các Tổng - Product of Sums):** Hàm là tích các Maxterms ($M_i$). Maxterm = tổng các biến (biến mang dấu phẩy nếu giá trị = 1, không phẩy nếu = 0).
  
### Phương pháp giải (Methods)
- **Rút gọn hàm logic bằng đại số:** Dùng các tính chất và định lý (đặc biệt là rút thừa số chung và luật nuốt) để làm biểu thức gọn đi. Phù hợp cho hàm tới 3 biến.
- **Triển khai hàm bằng NAND/NOR:** Dùng De Morgan hai lần: $F = \overline{\overline{F}}$ để biến cụm $(AB + CD)$ thành các cổng NAND/NOR hoàn toàn.

## 3. Rút gọn bằng Bìa Karnaugh (Karnaugh Maps)

### Phương pháp giải bài tập (Methods)
Bìa K cho 3, 4 hoặc 5 biến. Cấu trúc địa chỉ theo mã Gray (ví dụ $00, 01, 11, 10$ để đảm bảo tính kề nhau).

**Các bước tối thiểu hóa:**
1. Điền các số $1$ vào vị trí tương ứng nếu hàm dạng Tổng các Tích (hoặc điền $0$ nếu rút gọn dạng dạng POS). Điền dấu $x$ hoặc $d$ (Don't care) ở các tổ hợp biến không xác định.
2. Vòng lặp/Nhóm (Grouping): Khoanh vùng các ô chứa số $1$ kề nhau thành nhóm có kích thước là lũy thừa của 2 ($1, 2, 4, 8, 16,\dots$). Ưu tiên khoanh vòng lớn nhất có thể.
3. Kề ở cạnh mép: Phải nhớ rằng Bìa K cuộn tròn được, nghĩa là mép trái kề mép phải, mép trên kề mép dưới.
4. Khai thác Don't Care: Có thể bao gồm các ô $x$ nếu nó giúp vòng khoanh to lên (từ kích thước 2 lên 4, từ 4 lên 8).
5. Viết phương trình tối thiểu: Mỗi vòng khoanh sẽ tạo ra một P-term (tích). Các biến nào thay đổi giá trị trong nhóm đó thì bị gạch bỏ, giữ lại các biến không thay đổi. Tổng các vòng = Hàm SOP rút gọn.

## 4. Mạch Tổ Hợp (Combinational Circuits)

### Công thức & Khối cơ bản
- **Half Adder (Bộ cộng bán phần):** $S = A \oplus B$, $C_{out} = AB$
- **Full Adder (Bộ cộng toàn phần):** 
  - $S = A \oplus B \oplus C_{in}$
  - $C_{out} = AB + C_{in}(A \oplus B) = AB + BC_{in} + C_{in}A$
- **Multiplexer (Bộ dồn kênh - MUX $2^n \times 1$):** Chọn 1 trong $2^n$ đầu vào thông qua $n$ đường lựa chọn.
  - Phương trình MUX $4 \times 1$: $Y = \overline{S}_1\overline{S}_0 D_0 + \overline{S}_1 S_0 D_1 + S_1\overline{S}_0 D_2 + S_1 S_0 D_3$
- **Decoder / Demultiplexer (Giải mã / Phân kênh):** Chuyển mã nhị phân $n$-bit sang 1 trong $2^n$ đầu ra tương ứng bằng 1 (hoặc 0).

### Phương pháp giải bài tập (Methods)
- **Thực hiện hàm logic bằng MUX:** 
  - Nếu có hàm $n$ biến, dùng MUX $2^{n-1} \times 1$. Các biến MSB (từ bit $n-1$ đến $1$) kết nối vào chân lựa chọn nhận dạng. Biến LSB cuối cùng (ví dụ $C$) kết nối vào ngõ vào D (bằng 0, 1, C, hoặc $\overline{C}$) dựa trên so sánh bảng chân trị của MUX với bảng mức của hàm cần tổng hợp.
- **Thực hiện hàm logic bằng Decoder:** 
  - Đưa ngõ vào của hàm qua cọc nhập phân phối của Decoder. Đầu ra của Decoder sẽ xuất hiện các số hạng Minterms. Dùng 1 cổng $OR$ gộp những Minterms cần thiết để tạo thành hàm SOP.

## 5. Mạch Tuần Tự (Sequential Circuits)

### Công thức Trạng thái (Flip-Flops Characteristic Equations)
Các linh kiện nhớ 1 bit, đồng bộ qua xung nhịp Clock (CLK). 
- **RS Flip-Flop:** 
  $Q(t+1) = S + \overline{R}Q(t)$ *(Điều kiện ràng buộc: $SR = 0$)*
- **D Flip-Flop:** 
  $Q(t+1) = D$ 
  (Delay FF - Lưu giữ giá trị đầu vào sau một xung nhịp).
- **JK Flip-Flop:** 
  $Q(t+1) = J\overline{Q(t)} + \overline{K}Q(t)$
  (Khi $J=1, K=1$ FF lật trạng thái Toggle: $Q_{next} = \overline{Q}$). 
- **T Flip-Flop:** 
  $Q(t+1) = T \oplus Q(t)$
  (Chuyên dùng làm bộ đếm/chia tần số).

### Bảng phụ trợ kích (Excitation Tables)
Xác định ngõ vào FF cần thiết để đưa trạng thái $Q_n$ sang $Q_{n+1}$:
- Chuyển $0 \rightarrow 0$: JK = $0x$, T = $0$, D = $0$, RS = $0x$.
- Chuyển $0 \rightarrow 1$: JK = $1x$, T = $1$, D = $1$, RS = $10$.
- Chuyển $1 \rightarrow 0$: JK = $x1$, T = $1$, D = $0$, RS = $01$.
- Chuyển $1 \rightarrow 1$: JK = $x0$, T = $0$, D = $1$, RS = $x0$.
*(Trong đó $x$ là Don't care)*

### Phương pháp thiết kế hệ tuần tự (Đồng bộ) - Kỹ thuật máy trạng thái:
1. **Mô hình hóa (State Diagram):** Vẽ đồ thị Trạng thái dựa trên mô tả bài toán. Gán trạng thái thành giá trị nhị phân.
2. **Bảng chuyển trạng thái (State/Transition Table):** Bảng có cột Trạng Thái Hiện Tại, Inputs, Trạng Thái Kế Tiếp, và Outputs.
3. **Bảng kích thích FF (Excitation Table):** Thêm các cột tín hiệu kích (Kích $J, K, D$ hay $T$) theo loại FF sử dụng. 
4. **Karnaugh Map:** Thiết lập phương trình logic tối thiểu cho các ngõ vào của từng FF và đầu ra Output.
5. **Vẽ sơ đồ mạch:** Nối các cổng logic và FF lại với nhau.

### Phương pháp phân tích mạch tuần tự:
1. Viết biểu thức phương trình đại số kích tại các cổng FF (vd. $J_1 = QA, K_1=1...$).
2. Viết biểu thức Output.
3. Dùng phương trình đặc trưng của FF để tìm được phương trình cho Trạng Thái Kế Tiếp (Next State).
4. Lập bảng chân trị thể hiện sự biến hoán mạch rồi diễn ra đồ thị vòng xoay trạng thái (State Diagram).

## 6. Chuyển Đổi Tương Tự & Số (ADC / DAC)
Thông thường thuộc chương cuối.

### DAC (Digital to Analog Converter)
- **Mạng R-2R:** Phổ biến nhất cho DAC ICs.
- Vout tỉ lệ với mã nhị phân: $V_{out} = - V_{ref} \times \left(\frac{D_n}{2^n}\right)$ trong đó chữ số mã nhị phân $D = d_{n-1} \cdot 2^{n-1} + \dots + d_0$.
- Bước lượng tử hóa (Lọc phân giải, Resolution): $\Delta V = \frac{V_{FS}}{2^n} = 1 \text{ LSB}$.
  
### ADC (Analog to Digital Converter)
Chuyển điện áp về mặt số bằng lấy mẫu.
- **Flash ADC:** Quá trình chuyển rất nhanh do dùng bộ so sánh song song nhưng rườm rà ($2^n - 1$ bộ Amp. so sánh). 
- **Xấp xỉ xú liên tiếp (Successive Approximation ADC - SAR ADC):** Hội tụ trong n xung nhịp với n Bit.
- **Dual-slope ADC (Tích phân):** Chậm nhất, nhưng độ chính xác rất cao, miễn nhiễm với nhiễu.
- **Lỗi lượng tử (Quantization Error):** Sai số cao nhất là $\pm \frac{1}{2}$ LSB.
