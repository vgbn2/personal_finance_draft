## 1. Math & Physics Foundations
**Directory**: `Math_Physics/`

### Key Concepts
- **Brownian Motion ($W_t$)**: The continuous random walk. Fundamental to Physics (particle diffusion) and Finance (asset prices).
- **Itô's Lemma**: The Chain Rule for Stochastic Calculus.
  $$ df(t, X_t) = \left( \frac{\partial f}{\partial t} + \mu_t \frac{\partial f}{\partial x} + \frac{1}{2} \sigma_t^2 \frac{\partial^2 f}{\partial x^2} \right) dt + \sigma_t \frac{\partial f}{\partial x} dW_t $$

### Physics Connection
- **Heat Equation**: 
  $$ \frac{\partial u}{\partial t} = \alpha \nabla^2 u $$
  This describes how heat diffuses over time. It is mathematically isomorphic to the Black-Scholes equation after a coordinate transformation.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulation of Geometric Brownian Motion (Stock Price Model)
def simulate_gbm(S0, mu, sigma, T, N, M):
    dt = T/N
    t = np.linspace(0, T, N+1)
    S = np.zeros((N+1, M))
    S[0] = S0
    
    for i in range(1, N+1):
        dW = np.random.normal(0, np.sqrt(dt), M)
        S[i] = S[i-1] * np.exp((mu - 0.5 * sigma**2) * dt + sigma * dW)
    
    return t, S

# Params
S0 = 100
mu = 0.05
sigma = 0.2
T = 1.0
N = 252
M = 5

# Run
t, S = simulate_gbm(S0, mu, sigma, T, N, M)

# Plot
plt.figure(figsize=(10,6))
plt.plot(t, S)
plt.title('Geometric Brownian Motion: Stock Price Simulation')
plt.xlabel('Time (Years)')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```

## 2. Finance Applications
**Directory**: `Finance/`

### Black-Scholes Model
The most famous application of the Heat Equation in Finance.
$$ \frac{\partial V}{\partial t} + \frac{1}{2}\sigma^2 S^2 \frac{\partial^2 V}{\partial S^2} + rS \frac{\partial V}{\partial S} - rV = 0 $$

### Kelly Criterion
Optimal betting size to maximize geometric growth rate (Information Theory application in Finance).
$$ f^* = \frac{p(b+1) - 1}{b} $$

### Filtering vs. Smoothing
- **Telecom**: Removing thermal noise from a radio signal (Low-pass filter).
- **Finance**: Removing "market noise" to find the trend (Moving Average, Kalman Filter).

### Shannon Capacity
$$ C = B \log_2(1 + \frac{S}{N}) $$
Max theoretical data rate. In finance, this relates to the maximum rate of wealth growth (Kelly Criterion) given the "signal" (edge) and "noise" (volatility).
