- **Bộ nhớ ngắn hạn**: Bộ nhớ tạm thời lưu trữ thông tin cần thiết cho quá trình xử lý thần kinh.
- **TDNN**: Time delay neural network - Mạng nơ-ron với thời gian trễ, mở rộng đầu vào bằng N khung tín hiệu tiếng nói.
- **Matched Filter**: Mạch lọc tương hợp - Cấu trúc kết hợp với mạng nơ-ron truyền thống để giải quyết tính chất động của tín hiệu tiếng nói.
- **Độ tương đồng**: Đƣợc xác định bằng mô hình âm và xác suất tiên nghiệm.
- **Mô hình âm (acoustic model)**: Sử dụng để xác định độ tương đồng.
- **âm vị (phone)**: Một đơn vị âm thanh cơ bản trong một ngôn ngữ.
- **N-gram**: Một mô hình ngôn ngữ trong đó xác suất của một từ phụ thuộc vào N-1 từ trước đó.
- **Âm vị cơ sở**: Các âm vị cơ sở (Q) là một dãy các phát âm của từ trong mô hình nhận dạng.
- **HMM**: Mô hình Markov ẩn mật độ liên tục (HMM) là một mô hình được sử dụng để biểu diễn các âm cơ sở.
- **Cực đại kỳ vọng**: A method for estimating model parameters by maximizing the expected likelihood of the observed data.
- **Mạng nơ-ron nhân tạo**: Một mô hình máy tính được thiết kế để mô phỏng cấu trúc và hoạt động của bộ não con người.
- **Hàm phi tuyến**: Một hàm không tuyến tính, cho phép xử lý thông tin phi tuyến tính.
- **Quá trình Markov**: Một quá trình ngẫu nhiên trong đó tương lai của quá trình không phụ thuộc vào quá khứ.
- **Chuỗi Markov**: Một chuỗi Markov là một chuỗi Markov với giá trị nguyên và thời gian rời rạc.
- **Hidden Markov Model (HMM)**: A statistical model characterized by the number of states (N), the number of distinct observations (M), transition probability distribution (P), emission probability distribution (B), and initial state distribution (π).
- **State Transition Probability**: The probability of transitioning from one hidden state to another, represented by the distribution P, where Pr[i→j] represents the probability of moving from state i to state j.
