## <a id="chapter-1"></a>Chapter 1: Transistor Amplifiers

### 1.1 BJT Operating Point (Q-Point)
To use a transistor as an amplifier, it must be biased in the **Active Region**.
$$ I_C = \beta I_B $$
$$ V_{CE} = V_{CC} - I_C R_C $$

**Variables:**
- $I_C, I_B$: Collector and Base currents (Amps).
- $\beta$: DC Current Gain ($h_{FE}$).
- $V_{CE}$: Collector-Emitter Voltage.

**When to use:**
- Designing bias circuits (Fixed Bias, Voltage Divider Bias) to ensure the transistor stays on but not saturated.

### 1.2 Transistor IV Characteristics


```python
import matplotlib.pyplot as plt
import numpy as np

def plot_bjt_curves():
    v_ce = np.linspace(0, 20, 100)
    i_b_values = [10e-6, 20e-6, 30e-6, 40e-6, 50e-6] # 10uA to 50uA
    beta = 100
    
    plt.figure(figsize=(8, 5))
    
    for i_b in i_b_values:
        # Simplified Active region model with slight Early effect
        early_voltage = 100
        i_c = beta * i_b * (1 + v_ce / early_voltage)
        # Saturation region modeling
        i_c[v_ce < 0.2] = i_c[v_ce < 0.2] * (v_ce[v_ce < 0.2] / 0.2)
        
        plt.plot(v_ce, i_c * 1000, label=f'$I_B = {i_b*1e6:.0f} \mu A$')
        
    plt.title('BJT Output Characteristics ($I_C$ vs $V_{CE}$)')
    plt.xlabel('$V_{CE}$ (Volts)')
    plt.ylabel('$I_C$ (mA)')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

plot_bjt_curves()
```

## <a id="chapter-2"></a>Chapter 2: Operational Amplifiers (Op-Amps)

### 2.1 Inverting Amplifier
$$ A_v = \frac{V_{out}}{V_{in}} = -\frac{R_f}{R_{in}} $$

**Variables:**
- $A_v$: Voltage Gain.
- $R_f$: Feedback Resistor.
- $R_{in}$: Input Resistor.

**When to use:**
- To amplify and invert a signal. Input impedance is $R_{in}$.

### 2.2 Non-Inverting Amplifier
$$ A_v = 1 + \frac{R_f}{R_{g}} $$

**When to use:**
- High input impedance amplification. Buffer if $R_f=0$.


```python
from scipy import signal

def plot_bode_response():
    # Model of a standard Op-Amp (e.g., 741) with dominant pole
    # Open loop gain Ao = 10^5 (100 dB), Bandwidth = 10 Hz -> GBP = 1 MHz
    sys = signal.TransferFunction([1e5], [1/(2*np.pi*10), 1])
    w, mag, phase = signal.bode(sys)
    
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    plt.semilogx(w / (2*np.pi), mag, color='blue')    # Bode magnitude plot
    plt.title('Op-Amp Open Loop Gain')
    plt.ylabel('Magnitude (dB)')
    plt.xlabel('Frequency (Hz)')
    plt.grid(True, which="both", alpha=0.3)
    
    plt.subplot(1, 2, 2)
    plt.semilogx(w / (2*np.pi), phase, color='orange')  # Bode phase plot
    plt.title('Op-Amp Phase Response')
    plt.ylabel('Phase (degrees)')
    plt.xlabel('Frequency (Hz)')
    plt.grid(True, which="both", alpha=0.3)
    plt.tight_layout()
    plt.show()

plot_bode_response()
```

## <a id="chapter-3"></a>Chapter 3: Harmonic Oscillators

### 3.1 Resonance Frequency (LC Tank)
Condition for oscillation (Barkhausen): Loop Gain $\beta A = 1$.
$$ f_0 = \frac{1}{2\pi \sqrt{LC}} $$

**Variables:**
- $f_0$: Resonant Frequency (Hz).
- $L$: Inductance (Henry).
- $C$: Capacitance (Farad).

**When to use:**
- Designing Colpitts or Hartley oscillators for radio frequencies.


```python
def calc_resonance(L, C):
    '''Calculates resonance frequency f0 for LC circuit.'''
    return 1 / (2 * np.pi * np.sqrt(L * C))

L = 10e-6 # 10 uH
C = 1e-9  # 1 nF
f0 = calc_resonance(L, C)
print(f"Resonance Frequency for L={L*1e6}uH, C={C*1e9}nF: {f0/1e6:.2f} MHz")
```

## <a id="chapter-4"></a>Chapter 4: Pulse Circuits

### 4.1 555 Timer Astable Multivibrator
Generates a continuous square wave.
$$ f = \frac{1.44}{(R_1 + 2R_2)C} $$
$$ D = \frac{R_1 + R_2}{R_1 + 2R_2} \times 100\% $$

**Variables:**
- $f$: Frequency (Hz).
- $D$: Duty Cycle (%).
- $R_1, R_2, C$: External components.

**When to use:**
- Clock generation, LED flashers, Tone generation.
